```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Arun", "Priya", "Karthik", "Meena", "Rahul"],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT"],
    "Attendance": [82, 91, 76, 68, 88],
    "Python": [85, 92, 76, 65, 88],
    "SQL": [78, 88, 82, 70, 91],
    "Excel": [90, 95, 80, 72, 86]
}

df=pd.DataFrame(data)
df.sample(1)

print(df)
```

```python
#Python marks is at least 80
(df['Python'] >= 80)
```

```python

# Calculate the average of Python, SQL and Excel for filtered_students
df["Average_Marks"] = (df["Python"] + df["SQL"] + df["Excel"]) / 3

print(df[["Name", "Python", "SQL", "Excel", "Average_Marks"]])
```

```python
df["Average"] = df[["Python", "SQL", "Excel"]].mean(axis=1)

df["Results"] = df["Average"].apply(lambda x: "Pass" if x > 50 else "Fail")

print(df.to_string(index=False))
```

```python
df["Average"] = df[["Python", "SQL", "Excel"]].mean(axis=1)

df["Results"] = df["Average"].apply(lambda avg: "Pass" if avg > 50 else "Fail")

print(df.to_string(index=False))
```

```python
#Highest performing students list by Average

df.sort_values(by = "Average")
```

```python
# Highest performing students list by Attendance  - sort values

df.sort_values(by = "Attendance", ascending = False)
```

```python
#What is the average Python mark for each department ?
df.groupby("Department") ["Python"].mean()
```

```python
# Highest Python mark by department -group by
df.groupby("Department")["Python"].max()
```

```python
# Number of students per department -group by

df.groupby("Department")["Student_ID"].count()
```

```python
#Highest Python mark by department - sort / multiple group by

df.groupby("Department").agg(
    {
        "Student_ID": "count",
        "Python": "mean",
        "SQL": "mean",
        "Excel": "mean",
        "Attendance": "mean"
    }
)
```

```python
#chart insertion

import matplotlib.pyplot as plt

# Group by Department and calculate average Python marks
dep_py_avg = df.groupby("Department")["Python"].mean()

# Plot bar chart
dep_py_avg.plot(kind="bar", color="purple")

# Add chart labels and title
plt.title("Average Python Marks by Department")
plt.xlabel("Department")
plt.ylabel("Average Python Marks")

#Increase the size
plt.figure(figsize=(12,6))


# Show chart
plt.show()

```

```python
# Monthly Attendance
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
attendance = [82, 91, 76, 68, 88]
plt.plot(months, attendance, color="r", marker="*")
plt.xlabel("Months")
plt.ylabel("Attendance")
plt.title("Monthly Attendance")
plt.show()
```
