Examples

```python
# Find where the mark is greater than 80 
import numpy as np

# Define the 2D array
marks = np.array([
    [85, 78, 90],
    [92, 85, 98],
    [76, 82, 80]
])

# Condition: marks greater than 80
greater_than_80 = marks > 80

# Extract values
result = marks[greater_than_80]
# Find marks greater than or equal to 80
rows, columns = np.where(marks >= 80)

print("Indices:")
print("Rows:", rows)
print("Columns:", columns)

print("Original Marks:\n", marks)
print("Marks greater than or equal to 80:", marks[marks >= 80])

print("Sorted Marks:\n", np.sort(marks))
print((rows, columns))

print("Original Marks:\n", marks)
print("Marks greater than 80:", result)  # Gives exact values
np.sort(marks)[::-1]  # Row wise sort 
np.sort(marks,axis=1) # Column wise sort 


```
